```python
def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower_limit) |
        (data[column] > upper_limit)
    ]

    return len(outliers)


def cap_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    data[column] = data[column].clip(
        lower=lower_limit,
        upper=upper_limit
    )


outlier_columns = ['selling_price', 'km_driven']

print("Outliers Before Treatment")

for column in outlier_columns:
    count = detect_outliers_iqr(df, column)
    print(f"{column}: {count}")


for column in outlier_columns:
    cap_outliers_iqr(df, column)


print("\nOutliers After Treatment")

for column in outlier_columns:
    count = detect_outliers_iqr(df, column)
    print(f"{column}: {count}")


print("\nFINAL REPORT")
print("-" * 50)
print("Outliers treated in columns: selling_price, km_driven")
print("Dataset cleaning completed successfully")
print("Final Dataset Shape:", df.shape)
```

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, explained_variance_score, r2_score
```

```python
models = [LinearRegression, Ridge, Lasso, SVR, DecisionTreeRegressor, RandomForestRegressor]
mae = []
mse = []
rmse = []
evs = []
r_squared_score = []

for model in models:
    regressor = model().fit(x_train_scaled, y_train)
    y_pred = regressor.predict(x_test_scaled)
    print(y_pred)

    mae.append(mean_absolute_error(y_true=y_test, y_pred= y_pred))
    mse.append(mean_squared_error(y_true=y_test, y_pred= y_pred))
    rmse.append(np.sqrt(mean_squared_error(y_true=y_test, y_pred= y_pred)))
    evs.append(explained_variance_score(y_true=y_test, y_pred= y_pred))
    r_squared_score.append(r2_score(y_true=y_test, y_pred= y_pred))
    
```

```python
ml_models_df= pd.DataFrame({
    "models": ["LinearRegression", "Ridge", "Lasso", "SVR", "DecisionTreeRegressor", "RandomForestRegressor"],
    "mean_absolute_error":mae,
    "mean_squared_error":mse,
    "root_mean_squared_error":rmse,
    "explained_variance_score":evs,
    "r2_score": r_squared_score
    })

ml_models_df.set_index("models", inplace=True)

ml_models_df




```

```python
# HyperParameter Tuning
from sklearn.model_selection import GridSearchCV
```

```python
# define tune_params

params_grid = {
    LinearRegression:{},
    SVR:{
        'kernel':['rbf', 'linear'],
        'C':[0.1,1,10],
        'epsilon':[0.01,0.1,0.5],
    },
    DecisionTreeRegressor:{
        "max_depth":[None, 5,10,20],
        'min_samples_split':[2,5,10]   
    },
    RandomForestRegressor:{
        'n_estimators':[50,100,200,500],
        "max_depth":[None, 5,10,20],
        'min_samples_split':[2,5,10]
    },
    Ridge:{
        'alpha':[0.1,1.0,10.0]
    },
    
    Lasso:{
        'alpha':[0.001,0.01,0.1, 1.0]
    }
}

# metrics list

models = [LinearRegression, Ridge, Lasso, SVR, DecisionTreeRegressor, RandomForestRegressor]
mae = []
mse = []
rmse = []
evs = []
r_squared_score = []
best_models = []

for model in models:
    print(f"Tuning:{model.__name__}")
    if params_grid[model]:
        grid = GridSearchCV(model(), param_grid=params_grid[model],cv=5, scoring='r2', n_jobs=-1)
        grid.fit(x_train_scaled, y_train)
        regressor=grid.best_estimator_
        print(f"Best Parameters: {grid.best_params_}")
    else:
        regressor = model().fit(x_train_scaled, y_train)
    y_pred = regressor.predict(x_test_scaled)
    print(y_pred)

    mae.append(mean_absolute_error(y_true=y_test, y_pred= y_pred))
    mse.append(mean_squared_error(y_true=y_test, y_pred= y_pred))
    rmse.append(np.sqrt(mean_squared_error(y_true=y_test, y_pred= y_pred)))
    evs.append(explained_variance_score(y_true=y_test, y_pred= y_pred))
    r_squared_score.append(r2_score(y_true=y_test, y_pred= y_pred))
    best_models.append(regressor)
```

```python
ml_models_df= pd.DataFrame({
    "models": ["LinearRegression", "Ridge", "Lasso", "SVR", "DecisionTreeRegressor", "RandomForestRegressor"],
    "mean_absolute_error":mae,
    "mean_squared_error":mse,
    "root_mean_squared_error":rmse,
    "explained_variance_score":evs,
    "r2_score": r_squared_score
    })

ml_models_df.set_index("models", inplace=True)

ml_models_df




```
